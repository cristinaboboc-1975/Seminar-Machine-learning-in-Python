### 📝 Exercițiu — erori de tip (type errors)

Rulați, pe rând, fiecare dintre următoarele expresii, folosind:

```python
x = 3
y = '4'
z = '5.6'
```

- `x + y`
- `x + int(y + z)`
- `str(x) + int(y)`
- `str(x, y) + z`

Pentru fiecare, identificați **de ce** apare eroarea (sau, dacă nu apare, ce rezultat se obține).

````python
x = 3
y = '4'
z = '5.6'

x + y   # ❗ TypeError: nu poți aduna direct un int (x) cu un str (y)
````

**Explicații:**

- `x + y` → **eroare**: `x` este `int`, `y` este `str`; Python nu adună automat tipuri diferite.
- `x + int(y + z)` → **eroare**: `y + z` concatenează șirurile la `'45.6'`, iar `int('45.6')` eșuează pentru că șirul conține un punct zecimal (`int()` nu acceptă zecimale).
- `str(x) + int(y)` → **eroare**: `str(x)` este `'3'` (text), iar `int(y)` este `4` (număr); nu poți concatena text cu un număr folosind `+`.
- `str(x, y) + z` → **eroare**: `str()` nu acceptă doi parametri de acest fel — sintaxa este greșită.

Încearcă variantele corecte, de exemplu: `str(x) + y` sau `x + int(y)`.

## 7. Vectori, șiruri, tupluri și liste

Un **array** (vector, din `numpy`) conține o secvență de valori:

- toate elementele unui vector au, de regulă, același tip;
- operațiile aritmetice se aplică individual fiecărui element;
- când doi vectori sunt adunați, trebuie să aibă aceeași dimensiune — elementele corespunzătoare se adună între ele;
- o coloană dintr-un set de date este, în esență, un vector.

Un **range** (șir) este un vector de numere consecutive:
- `np.arange(end)` — generează numerele de la 0 până la `end - 1`.

````python
# crearea unui vector folosind funcția numpy arange
my_array = np.arange(2, 8)
my_array
````

````python
# crearea unei liste — folosind paranteze pătrate
# spre deosebire de vectori, listele pot conține valori de tipuri diferite
my_list = [1, 2, 3, 4]
my_list
````

````python
my_list1 = [1, 2, 3, 4, 'c']
my_list1
````

````python
print(sum(my_array))
print(my_array.sum())
````

⚠️ Următoarea celulă va genera o eroare: `sum()` nu poate aduna elemente de tipuri diferite (numere și text) dintr-o listă.

````python
print(sum(my_list1))   # ❗ TypeError
````

````python
print(sum(my_list))   # aici funcționează, toate elementele sunt numere
````

````python
# adunarea a doi vectori — elementele corespunzătoare se adună
# (cei doi vectori trebuie să aibă aceeași dimensiune!)
my_array1 = np.arange(5, 11)
print(my_array1)
my_array + my_array1
````
